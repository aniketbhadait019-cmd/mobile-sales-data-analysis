# **Univariate Analysis**

**check Payment Method distribution**

```python
plt.figure(figsize=(7,3))
sns.histplot(mobile_sales['Payment_Method'],color='teal')
plt.title('Payment_Method Distribution',fontweight='bold')
plt.xlabel('Payment_Method',color='k',fontsize=10,fontweight='bold')
plt.ylabel('Payments',fontweight='bold')
plt.show()
```

 **Check Mobile Unit Price**

```python
plt.figure(figsize=(8,5))
sns.histplot(data=mobile_sales,x='Unit_Price_INR',bins=30,color='olivedrab')
plt.title("Distribution of Mobile Unit Price",fontweight='bold')
plt.xlabel("Unit Price (INR)")
plt.ylabel("Frequency")
plt.show()
```

**Check Rating**

```python
plt.figure(figsize=(8,5))
sns.violinplot(data=mobile_sales,x='Customer_Rating').axes.set_facecolor('olive')
plt.title("Customer_Rating Distribution",fontweight='bold')
plt.xlabel("Customer_Rating",fontweight='bold')
plt.show()
```

**Revenue Distribution**

```python
plt.figure(figsize=(8,5))
sns.kdeplot(data=mobile_sales,x='Revenue_INR',color='k',fill=False).axes.set_facecolor('lightgreen')
plt.title("Revenue Distribution",fontweight='bold')
plt.xlabel("Revenue (INR)",fontsize='12',fontweight='bold')
plt.ylabel("Density",fontsize='12',fontweight='bold')
plt.show()
```

```python
price_segment_count = mobile_sales['Price_Segment'].value_counts()
plt.figure(figsize=(7,7))
plt.pie(price_segment_count.values,labels=price_segment_count.index,autopct='%10.1f%%',startangle=90)
plt.title("Price Segment Distribution",color='maroon')
plt.show()
```

**Most Mobile Network**

```python
plt.figure(figsize=(8,5))
network_counts = mobile_sales['Network'].value_counts()
plt.bar(x=network_counts.index, height=network_counts.values, color='khaki')
plt.title(" Most Mobile Network",fontweight='bold')
plt.xlabel("Network",fontweight='bold')
plt.ylabel("Count",fontweight='bold')
plt.gca().set_facecolor('gray')
plt.xticks(rotation=0)
plt.show()
```

# **Feature Engineering**

**Age Group**

```python
mobile_sales['Age_Group'] = pd.cut(
    mobile_sales['Customer_Age'],
    bins=[0, 18, 30, 45, 60, 100],
    labels=[
        'Below 18',
        '18-30',
        '31-45',
        '46-60',
        '60+'])
```

**High / Low Rating**

```python
mobile_sales['Rating_Category'] = np.where(mobile_sales['Customer_Rating'] >= 4,'High Rating','Low Rating')
```

**Price Category**

```python
mobile_sales['Price_Category'] = pd.cut(
    mobile_sales['Unit_Price_INR'],
    bins=[0, 10000, 20000, 40000, 1000000],
    labels=['Budget','Mid Range','Premium','Luxury'])
```

# Key Insights

1.Mobile sales vary across different years and months, showing changes in sales activity over time.
2.Brand performance is different, with some brands contributing more sales and revenue than others.
3.Online and Offline channels both contribute to mobile sales, and their revenue contribution can be compared using the channel analysis.
4.Different price segments have different sales volumes. This helps identify which price range has greater customer demand.
5.Revenue differs across months, which helps identify high-revenue and low-revenue periods.
6.Customer demographics influence sales patterns. Customer gender, age group, and payment method can be used to understand purchasing behavior.
7.Customer ratings show variation among customers, which provides useful information about customer satisfaction.
8.Discount percentage and revenue can be analyzed together to understand how discounts are associated with sales revenue.
9.Mobile network preferences such as 4G and 5G differ across customers, providing useful information about network demand.
10.The analysis identified missing values and duplicate records, which were handled during the data-cleaning stage to improve data quality.

# **Conclusion**

The Mobile Sales EDA provides a clear understanding of sales performance, revenue, customer behavior, brands, pricing, discounts, payment methods, and sales channels. The analysis shows that mobile sales are influenced by several factors such as brand, price segment, month, channel, customer characteristics, and purchasing preferences.

The visualizations and statistical analysis make it easier to identify sales trends, high-performing categories, customer patterns, and revenue differences. Overall, this EDA can help a business understand its mobile sales data and use the identified patterns for better sales planning, marketing strategies, pricing decisions, and customer analysis.
